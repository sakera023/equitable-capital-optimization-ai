# Example 1 — Capital Readiness Analysis

This notebook reproduces the core predictive workflow: generate privacy-safe synthetic business records, train the baseline model, inspect holdout metrics, and review capital-readiness scores.

> **Research use only.** The synthetic data and model must not be used for real credit, lending, investment, or eligibility decisions.


```python
import pandas as pd
import plotly.express as px

from equitable_capital import (
    generate_synthetic_startups,
    global_feature_importance,
    train_model,
)
```

## Generate a reproducible synthetic dataset


```python
data = generate_synthetic_startups(n=1500, seed=42)
data.head()
```

## Train the model and inspect holdout performance


```python
result = train_model(data)
pd.DataFrame(
    [{"metric": name, "value": value} for name, value in result.metrics.items()]
)
```

## Review scored businesses


```python
scored = result.scored_data
scored[[
    "startup_id",
    "state",
    "industry",
    "requested_capital",
    "predicted_success_probability",
    "capital_readiness_score",
    "underserved_context_index",
]].head(10)
```

## Visualize requested capital versus readiness


```python
fig = px.scatter(
    scored,
    x="requested_capital",
    y="capital_readiness_score",
    color="underserved_context_index",
    hover_data=["startup_id", "state", "industry"],
    title="Synthetic Capital Need vs. Predicted Capital Readiness",
)
fig.show()
```

## Inspect global feature importance

Feature importance is model-specific and does not imply causality.


```python
importance = global_feature_importance(result).head(12)
importance
```

```python
fig = px.bar(
    importance.sort_values("importance"),
    x="importance",
    y="feature",
    orientation="h",
    title="Top Global Feature Importances",
)
fig.show()
```

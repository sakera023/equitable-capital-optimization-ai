# Example 4 — Official U.S. Small-Business Data Context

This notebook demonstrates the public-data layer using authoritative aggregate datasets from the U.S. Small Business Administration Office of Advocacy.

Internet access is required because the package resolves the current official workbook from the SBA open-data catalog at runtime.


```python
import pandas as pd

from equitable_capital import (
    PUBLIC_DATASETS,
    get_sba_dataset_metadata,
    load_sba_public_workbook,
)
```

## Review the supported official sources


```python
pd.DataFrame(
    [
        {
            "key": key,
            "label": value["label"],
            "landing_page": value["landing_page"],
        }
        for key, value in PUBLIC_DATASETS.items()
    ]
)
```

## Resolve live SBA catalog metadata


```python
dataset_key = "sba_state_2025"
metadata = get_sba_dataset_metadata(dataset_key)
pd.Series(metadata)
```

## Download the current official workbook


```python
metadata, sheets = load_sba_public_workbook(dataset_key)
[(name, frame.shape) for name, frame in sheets.items()]
```

## Inspect the first available worksheet


```python
sheet_name = next(iter(sheets))
public_frame = sheets[sheet_name]
print(f"Sheet: {sheet_name} | Shape: {public_frame.shape}")
public_frame.head(20)
```

## Inspect available fields and numeric measures


```python
print("Columns:")
for column in public_frame.columns:
    print(" -", column)

numeric_columns = public_frame.select_dtypes(include="number").columns.tolist()
print("\nNumeric columns:", numeric_columns)
```

## Research-use note

These public statistics are aggregate contextual evidence. They are deliberately kept separate from the synthetic applicant-level prediction model and should not be presented as validation of individual funding predictions.

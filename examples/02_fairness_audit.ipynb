# Example 2 — Fairness and Opportunity Audit

This notebook evaluates how model outcomes differ between lower- and higher-structural-barrier contexts in the synthetic research dataset.

The audit is a diagnostic research tool, not a legal determination of discrimination or compliance.


```python
import pandas as pd
import plotly.express as px

from equitable_capital import (
    fairness_audit,
    generate_synthetic_startups,
    opportunity_gap,
    train_model,
)
```

```python
data = generate_synthetic_startups(n=1500, seed=42)
result = train_model(data)
scored = result.scored_data
```

## Baseline audit at a 0.50 selection threshold


```python
audit = fairness_audit(scored, threshold=0.50)
audit
```

## Opportunity-gap summary


```python
gap = opportunity_gap(scored)
pd.Series(gap, name="value")
```

## Sensitivity across several selection thresholds


```python
audits = []
for threshold in [0.40, 0.50, 0.60]:
    frame = fairness_audit(scored, threshold=threshold).copy()
    frame["threshold"] = threshold
    audits.append(frame)

threshold_audit = pd.concat(audits, ignore_index=True)
threshold_audit[[
    "threshold",
    "context_group",
    "n",
    "selection_rate",
    "avg_predicted_success",
    "avg_readiness_score",
    "selection_rate_ratio",
]]
```

```python
fig = px.line(
    threshold_audit,
    x="threshold",
    y="selection_rate",
    color="context_group",
    markers=True,
    title="Selection-Rate Sensitivity by Structural Context",
)
fig.show()
```

## Interpretation

Use these diagnostics to identify distributional differences that deserve further investigation. Do not infer causality from the structural-context index or from model-output differences alone.

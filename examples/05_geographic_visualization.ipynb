# Example 5 — Geographic Visualization

This notebook creates a U.S. state choropleth from the synthetic research dataset.

> The mapped values are synthetic research outputs, not measured state economic outcomes.


```python
import plotly.express as px

from equitable_capital import generate_synthetic_startups, summarize_synthetic_states, train_model
```

```python
data = generate_synthetic_startups(n=1500, seed=42)
result = train_model(data)
state_geo = summarize_synthetic_states(result.scored_data)
state_geo
```

## Map average Capital Readiness Score


```python
fig = px.choropleth(
    state_geo,
    locations="state",
    locationmode="USA-states",
    color="avg_readiness",
    scope="usa",
    hover_name="state_name",
    title="Synthetic State-Level Capital Readiness",
)
fig.update_layout(margin={"r": 0, "t": 55, "l": 0, "b": 0})
fig.show()
```

## Interpretation

Geographic color differences are descriptive research visualizations and do not establish causation, discrimination, or policy effectiveness.
